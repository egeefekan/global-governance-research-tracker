```python
import pandas as pd

url = "https://raw.githubusercontent.com/egeefekan/global-governance-research-tracker/refs/heads/main/data/sample_governance_data.csv"

df = pd.read_csv(url)

df
```

```python
df[["country", "gdp_per_capita"]].sort_values("gdp_per_capita", ascending=False)
```

## Early Observation

In this sample dataset, Germany has the highest democracy score, while Singapore has the highest GDP per capita. This suggests that higher GDP per capita does not automatically mean a higher democracy score.

```python
df.plot(x="country", y="democracy_score", kind="bar")
```

```python
import matplotlib.pyplot as plt

democracy_sorted = df.sort_values("democracy_score", ascending=False)

democracy_sorted.plot(
    x="country",
    y="democracy_score",
    kind="bar",
    legend=False
)

plt.title("Democracy Score by Country")
plt.xlabel("Country")
plt.ylabel("Democracy Score")
plt.xticks(rotation=45)
plt.show()
```

```python
gdp_sorted = df.sort_values("gdp_per_capita", ascending=False)

gdp_sorted.plot(
    x="country",
    y="gdp_per_capita",
    kind="bar",
    legend=False
)

plt.title("GDP per Capita by Country")
plt.xlabel("Country")
plt.ylabel("GDP per Capita")
plt.xticks(rotation=45)
plt.show()
```

## Early Observation

In this sample dataset, Germany has the highest democracy score, while Singapore has the highest GDP per capita. This suggests that higher GDP per capita does not automatically mean a higher democracy score.

Because this is a small sample dataset created for learning purposes, no serious conclusion should be drawn from it yet.

```python
internet_sorted = df.sort_values("internet_penetration", ascending=False)

internet_sorted.plot(
    x="country",
    y="internet_penetration",
    kind="bar",
    legend=False

)
plt.title("Internet Penetration per Country")
plt.xlabel("Country")
plt.ylabel("Internet Penetration")
plt.xticks(rotation=25)
plt.show()

```

## Internet Penetration Observation

This chart compares internet penetration across the selected countries. Singapore has the highest internet penetration rate in this sample dataset, while Italy has the lowest.

Because this dataset includes only five countries and was created for learning purposes, it should not be used to make serious conclusions.


## What I Learned

1. **What is a CSV file?**  
   A CSV file is a simple file format that stores table data as plain text. Each row represents a record, and commas separate the columns.

2. **What does pandas do?**  
   Pandas is a Python library used to read, organize, analyze, and manipulate data. In this project, I use pandas to read the CSV file and work with it as a table.

3. **What does `sort_values()` do?**  
   `sort_values()` sorts the rows of a dataset based on the values in a selected column. For example, I used it to sort countries from the highest to the lowest democracy score.

4. **What does a bar chart help us compare?**  
   A bar chart helps compare values across categories. In this project, it helps compare countries using indicators such as GDP per capita, democracy score, and internet penetration.
